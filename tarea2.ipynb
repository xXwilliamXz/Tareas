## Agrupar datos 
df.groupby()
* Con este código podemos agrupar columnas e incluso otros dataframes y aplicarles alguna acción, como en este caso con los promedios, conteos por categoría.

Para el groupby, escribimos:
df.groupby("categoría")[["columnas que querémos"]]

```python
#el primer () debe ser un dato categorico
df.groupby("col10")[["col1","col3","col4","col7","col8"]].mean()
```

```python
df.groupby(["col10"])[["col1","col3","col4","col7","col8"]].count()
```

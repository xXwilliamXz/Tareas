## Nivel 4 — Lambda + pivot table

16. Crea una variable que clasifique los pedidos según el costo total y después construye una pivot table con el costo total promedio por categoría y segmento.

```python
df16 = order_items.merge(products, on = "product_id", how= "left")
df16["costo_total"] = df16.apply(lambda fila: (fila["price"]+ fila["freight_value"]) , axis = 1)
df16["caro_o_barato"] = df16["costo_total"].apply(lambda x: "Caro" if x > 150 else "Barato")
df16.pivot_table(values="costo_total", index="product_category_name",columns="caro_o_barato", aggfunc="mean")
```

17. Calcula el porcentaje que representa el envío respecto al precio y crea una clasificación: "envío bajo" y "envío alto". Después compara ambas categorías mediante una pivot table.

```python
df17 = df16
df17["porcentaje_envio"] = df17.apply(lambda fila: (fila["freight_value"]/fila["price"]), axis = 1)
df17["clasif"] = df17["porcentaje_envio"].apply(lambda x: "envio_alto" if x> 0.3 else "envio_bajo")
df17.pivot_table(values="freight_value", index ="product_category_name", columns="clasif", aggfunc="mean")
```

18. Crea una variable que indique si el pedido tiene un precio superior al precio promedio de todos los productos. Después analiza esa variable mediante una pivot table.

```python
df18 = df16
promedio = df18["price"].mean()
df18["variable_promedio"] = df18["price"].apply(lambda x:"superior al promedio" if x>promedio else "inferior al promedio")
df18.pivot_table(values= "price", index="product_category_name", columns="variable_promedio", aggfunc="count")
```

22. Construye un pequeño reporte que responda: ¿qué categorías presentan comportamientos diferentes dependiendo del estado del pedido? Utiliza al menos dos pivot tables y una transformación realizada con lambda.

```python
df22 = df16.merge(orders, on="order_id", how="left")
df22 = df22[df22["order_status"].isin(["delivered", "canceled"])] #Aquí solo me quedo con los estatus entregado y cancelado
flete_prom = df22.pivot_table(values="freight_value", index="product_category_name", columns="order_status", aggfunc= lambda x: x.max()-x.min())
cancelaciones = df22.pivot_table(values="price", index="product_category_name", columns="order_status", aggfunc=lambda x: x.mean(), fill_value=0)

display(flete_prom, cancelaciones)
```

Gracias a las siguientes tablas podemos ver que para cada función hay una variacion dependiendo al estatus de la orden. En la primera tabla podemos apreciar el flete proemdio de cada categoria por estado, el cual nos sirve para ubicar fletes muy altos, si alguna categoria tuviese un costo de flete mas elevado en cancelado que entregado, nos indicaría que el cliente decide no comprar cuando ve el costo del envio xd.
Mientras tanto, en la segunda tabla podemos apreciar cuales son las categorias las cuales concentran mayor numero de cancelaciones o entregas efectivas, donde podemos ver que entre resultados distan bastante, con categorias donde no se ha cancelado ninguna orden.
